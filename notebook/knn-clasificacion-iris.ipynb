## KNN - K Vecinos Mas Cercanos
* Referencia: https://youtu.be/iWZpw6M0f-Q
* Dataset: IRIS

```python
import pandas as pd

# Cargamos dataset
df = pd.read_csv("../data/iris.csv")

df.info()
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

x = df[["SepalLengthCm", "SepalWidthCm"]]

# Codificamos etiquetas a numeros enteros para compatibilidad con mlxtend
le = LabelEncoder()
y = le.fit_transform(df["Species"])
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y)
```

```python
# Preprocesamiento - Standarizacion Datos
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(x_train)

x_train_std = scaler.transform(x_train)
x_test_std = scaler.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3, p=2, metric='minkowski')
```

```python
# Realizamos entrenamiento modelo knn
knn.fit(x_train_std, y_train)
```

```python
# Visualizamos Resultados
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions

ax = plot_decision_regions(x_test_std, y_test, clf=knn, legend=2)

# Mostrar nombres de especies en la leyenda
handles, _ = ax.get_legend_handles_labels()
ax.legend(handles, le.classes_, loc="upper left")

plt.xlabel("Sepalo largo CM")
plt.ylabel("Sepalo ancho CM")
plt.title("KNN aplicado en IRIS dataset")
plt.show()

print("Scores Modelo:", knn.score(x_test_std, y_test))
```
