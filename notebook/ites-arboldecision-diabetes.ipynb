```python
import pandas as pd

df = pd.read_csv("../data/diabetes.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# Verificar si hay clases desbalanceadas
df['Outcome'].value_counts()
```

```python
# Creacion del set de entrenamiento y prueba
from sklearn.model_selection import train_test_split

# x: entradas, y: salida
x = df.iloc[:,:-1]
y = df.iloc[:,-1]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42, stratify=y)
```

```python
# Imprimir tamaños resultantes
print("Tamaño set de entrenamiento: ", x_train.shape, y_train.shape)
print("Tamaño set de test: ", x_test.shape, y_test.shape)
```

```python
# Entrenamiento y visualizacion

from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(random_state=42)

modelo.fit(x_train, y_train)
```

```python
# Graficar el arbol
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt


plt.figure(figsize=(30,15))
plot_tree(modelo,
    feature_names=x_train.columns,
    class_names=["No Diabetes", "Diabetes"],
    filled=True,    # Colorear las hojas con la categoria mayoritaria
    rounded=True
)

# Guardar el arbol como archivo PNG
plt.savefig("arbol_clasificacion.png", dpi=700)
```

```python
print("Profundidad del arbol: ", modelo.get_depth())
print("Numero hojas del arbol: ", modelo.get_n_leaves())
```

## Metricas de Desemepño del Arbol

* TP (verdaderos positivos): diabeticos detectados correctamente como diabeticos.
* FP (falso positivo): sanos incorrectamente clasificados como diabeticos
* TN (verdaderos negativos): sanos dectectados correctamente como sanos
* FN (falsos negativo): diabeticos incorrectamente detectados como sanos 

Precision: ¿de todos los clasificados diabeticos que porcentaje es realmente diabetico?

Recall: ¿de todos los diabeticos que proporcion fue clasificada realmente como diabeticos?

F1-score: combina precision y recall en un unica metrica



```python
# Validacion cruzada
from sklearn.model_selection import StratifiedKFold, cross_val_score

# Crear particiones stratificadas
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Realizar validacion cruzada (k-fold)
scores = cross_val_score(modelo, x_train, y_train, cv=skf, scoring="f1_weighted")

# Imprmir resultados de validacion cruzada
print("F1-score ponderado entrenamiento:", scores.mean())

# Puntaje de set de prueba
score_test = modelo.score(x_test, y_test)
print("F1-score ponderado prueba:", score_test)
```
